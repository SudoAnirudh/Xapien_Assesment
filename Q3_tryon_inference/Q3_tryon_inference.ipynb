```python
# Environment Setup & Google Drive Mount
import os
import sys

def _setup_phase_path(phase_name):
    if os.path.exists('config.py'):
        p = os.path.abspath('.')
        if p not in sys.path: sys.path.insert(0, p)
        return p
    if os.path.exists(os.path.join(phase_name, 'config.py')):
        p = os.path.abspath(phase_name)
        if p not in sys.path: sys.path.insert(0, p)
        try: os.chdir(p)
        except Exception: pass
        return p
    for search_base in [os.getcwd(), os.path.dirname(os.getcwd()), '/content']:
        if os.path.exists(search_base):
            for root, dirs, files in os.walk(search_base):
                if 'config.py' in files and phase_name in root:
                    p = os.path.abspath(root)
                    if p not in sys.path: sys.path.insert(0, p)
                    try: os.chdir(p)
                    except Exception: pass
                    return p
    return os.path.abspath('.')

_setup_phase_path('Q3_tryon_inference')

try:
    from google.colab import drive
    drive.mount('/content/drive')
    DRIVE_MOUNTED = True
    CHECKPOINT_DIR = '/content/drive/MyDrive/xipl_tryon_checkpoints/Q3'
    os.makedirs(CHECKPOINT_DIR, exist_ok=True)
    print(f"[INFO] Google Drive mounted. Checkpoints DIR: {CHECKPOINT_DIR}")
except ImportError:
    DRIVE_MOUNTED = False
    CHECKPOINT_DIR = './checkpoints/Q3'
    os.makedirs(CHECKPOINT_DIR, exist_ok=True)
    print(f"[INFO] Local environment detected. Checkpoints DIR: {CHECKPOINT_DIR}")
```
